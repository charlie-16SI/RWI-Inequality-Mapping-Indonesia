```python
import geopandas as gpd
import pandas as pd

# 1. Muat data RWI
df_rwi = pd.read_csv("idn_relative_wealth_index.csv")
```

```python
gdf_rwi = gpd.GeoDataFrame(
    df_rwi,
    geometry=gpd.points_from_xy(df_rwi.longitude, df_rwi.latitude),
    crs="EPSG:4326",
)
```

```python
gdf_kab = gpd.read_file("gadm41_IDN_2.json")
```

```python
joined = gpd.sjoin(gdf_rwi, gdf_kab, how="inner", predicate="within")
```

```python
rwi_kabupaten = (
    joined.groupby(["NAME_1", "NAME_2"])["rwi"]
    .agg(rata_rata_rwi="mean", jumlah_titik="count")
    .reset_index()
)
```

```python
rwi_kabupaten.to_excel("rwi_per_kabupaten.xlsx", index=False)
```
